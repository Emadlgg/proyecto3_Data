# Encuesta de visitantes de Petén 2025

Análisis descriptivo de la muestra por cuotas de INGUAT. **Sin factor de expansión:** no representa totales ni proporciones de la población turística.

```python
from pathlib import Path
import sys
import pandas as pd
import plotly.express as px
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks": ROOT = ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from src.encuestas import leer_encuesta, porcentajes_muestra, satisfaccion_muestra, cobertura_preguntas
PATH = ROOT / "data/raw/Encuesta_visitante_Peten_2025.xlsx"
data, diccionario, nota = leer_encuesta(PATH)
print(f"Respuestas: {len(data)}; columnas: {len(data.columns)}")
display(data[["Lugar_Turistico", "tipo_visitante", "region", "p05", "p06"]].head())
```

## Cobertura y filtros

La ausencia de respuesta puede deberse a filtros del cuestionario. Comparar con el diccionario antes de interpretar los nulos.

```python
cobertura = cobertura_preguntas(data, diccionario)
display(cobertura.sort_values("porcentaje_observado").head(20))
```

## Perfil descriptivo

Porcentajes calculados sobre las respuestas válidas **dentro de cada subdestino y tipo de visitante**.

```python
perfil = porcentajes_muestra(data)
display(perfil.head(15))
genero = perfil.query("indicador == 'genero'")
fig = px.bar(genero, x="Lugar_Turistico", y="porcentaje_muestra", color="respuesta", barmode="group", facet_col="tipo_visitante", title="Género entre respuestas válidas de la muestra")
fig.show()
```

## Satisfacción (escala 1 a 5)

La media describe únicamente a quienes respondieron cada pregunta; no es una media poblacional.

```python
satisfaccion = satisfaccion_muestra(data, diccionario)
display(satisfaccion.query("variable == 'T_p65'"))
fig = px.bar(satisfaccion.query("variable == 'T_p65'"), x="Lugar_Turistico", y="media_muestra", color="tipo_visitante", barmode="group", range_y=[0,5], title="Satisfacción general (muestra)")
fig.show()
```

## Conclusión metodológica

Los resultados son descriptivos de una sola ola levantada por cuotas. No extrapolar a visitantes de Petén, no comparar como serie de tiempo ni mezclar porcentajes con los conteos de telefonía, Tikal o AIMM. Antes de integrar al dashboard, revisar cobertura y posibles codificaciones particulares de cada pregunta.
